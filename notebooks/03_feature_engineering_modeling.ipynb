# 3. Feature Engineering e Modelagem Preditiva

Neste notebook vamos criar variáveis inteligentes e treinar os modelos de Machine Learning (Logistic Regression, Random Forest, XGBoost).

```
import sys
sys.path.append('../src')
from data_processing import process_pipeline
from model_training import prepare_split, train_and_evaluate
import pandas as pd
```

## 3.1 Geração de Features e Preparação

```
df_final, df_feat = process_pipeline('../data/raw/dataset_churn_investidores.xlsx', is_train=True)
X_train, X_test, y_train, y_test = prepare_split(df_final)
print('Shape do treino:', X_train.shape)
```

## 3.2 Treinamento e Avaliação

```
models, results = train_and_evaluate(X_train, X_test, y_train, y_test)
pd.DataFrame(results).T
```

## 3.3 Feature Importance (XGBoost)

```
best_model = models['XGBoost']
importances = pd.DataFrame({'Feature': X_train.columns, 'Importance': best_model.feature_importances_})
importances = importances.sort_values(by='Importance', ascending=False)
import plotly.express as px
fig = px.bar(importances.head(10), x='Importance', y='Feature', orientation='h', title='Top 10 Features - XGBoost')
fig.gca().invert_yaxis()
fig.show()
```
