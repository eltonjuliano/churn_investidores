# 2. Análise Exploratória de Dados (EDA)

Aprofundando no comportamento de churn dos investidores.

```
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel('../data/raw/dataset_churn_investidores.xlsx')
```

## 2.1 Taxa Geral de Churn

```
churn_rate = df['churn'].value_counts(normalize=True) * 100
fig = px.pie(values=churn_rate, names=['Retido', 'Churn'], title='Distribuição de Churn Geral')
fig.show()
```

## 2.2 Churn por Perfil de Investidor

```
fig = px.histogram(df, x='perfil_investidor', color='churn', barmode='group', title='Churn por Perfil do Investidor')
fig.show()
```

## 2.3 Relação: Dias sem Login vs Churn

```
fig = px.box(df, x='churn', y='dias_sem_login', color='churn', title='Dias Sem Login vs Churn')
fig.show()
```

## 2.4 Matriz de Correlação (Numéricas)

```
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
corr = df[num_cols].corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap='coolwarm', annot=False)
plt.title('Correlação de Variáveis Numéricas')
plt.show()
```
